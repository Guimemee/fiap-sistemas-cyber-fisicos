**Título:** Arquitetura de Controle e Atuação para Exoesqueleto Assistivo Upper Limb

> **Enunciado Original:**
> Dimensionar o sistema de acionamento, motores e malha de controle de torque e posição para um exoesqueleto robótico de membro superior (Exo Upper Limb) projetado para redução de esforço ergonômico no levantamento e sustentação de ferramentas pesadas na indústria automotiva. Calcular o torque necessário na articulação do cotovelo para carga de 5 kg.

### Formulação Matemática Formal
$$\tau_{\text{cotovelo}} = m_{\text{carga}} \cdot g \cdot L_{\text{antebraço}} + m_{\text{braço}} \cdot g \cdot \frac{L_{\text{antebraço}}}{2}, \quad P_{\text{mec}} = \tau \cdot \omega$$


### Passo 1: Parâmetros Biomecânicos e Geometria do Membro
Comprimento médio do antebraço L = 0,35 m; massa da carga de ferramenta M = 5,0 kg; aceleração gravitacional g = 9,81 m/s²; massa própria do segmento anatômico m_b = 1,8 kg.


### Passo 2: Cálculo do Torque Estático Máximo (Braço a 90°)
Torque da carga: \tau_c = 5,0 \times 9,81 \times 0,35 = 17,1675\text{ N}\cdot\text{m}. Torque do próprio antebraço: \tau_b = 1,8 \times 9,81 \times 0,175 = 3,090\text{ N}\cdot\text{m}. Torque total estático: \tau = 17,17 + 3,09 = 20,26\text{ N}\cdot\text{m}.


### Passo 3: Margem Dinâmica de Aceleração e Inércia
Considerando aceleração angular rápida de \alpha = 2,5\text{ rad/s}^2 e inércia I = 0,65\text{ kg}\cdot\text{m}^2: \tau_{din} = 20,26 + (0,65 \times 2,5) = 20,26 + 1,625 = 21,885\text{ N}\cdot\text{m}.


### Passo 4: Seleção do Atuador e Redutor Harmonic Drive
Servomotor BLDC sem carcaça (frameless) com redutor de engrenagem harmônica (relação N = 50:1 e rendimento \eta = 82%). Torque no eixo do motor: \tau_{mot} = \frac{21,89}{50 \times 0,82} \approx 0,534\text{ N}\cdot\text{m}.


### Passo 5: Conclusão de Alívio Ergonômico
O atuador fornece 85% do torque requerido, reduzindo a solicitação do músculo bíceps braquial do operador em mais de 70%, prevenindo lesões por esforço repetitivo (LER/DORT).


```python
import numpy as np
import matplotlib.pyplot as plt

angulo_deg = np.linspace(0, 140, 200) # Angulo de flexao do cotovelo (0 = estendido)
ang_rad = np.radians(angulo_deg)
# Torque estatico em funcao do angulo (maximo a 90 graus)
tau_sem_exo = (5.0 * 9.81 * 0.35 + 1.8 * 9.81 * 0.175) * np.sin(ang_rad)
tau_com_exo = tau_sem_exo * 0.25 # Exo alivia 75%
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0f2526')
ax.set_facecolor('#081617')
ax.plot(angulo_deg, tau_sem_exo, color='#ef4444', lw=2.5, label='Esforço Biomecânico Sem Exoesqueleto')
ax.plot(angulo_deg, tau_com_exo, color='#14b8a6', lw=2.5, label='Esforço Residual do Usuário Com Exoesqueleto')
ax.fill_between(angulo_deg, tau_com_exo, tau_sem_exo, color='#14b8a6', alpha=0.2, label='Alívio Fornecido pelo Atuador (75%)')
ax.set_title('Torque Articular no Cotovelo vs. Ângulo de Flexão', color='#ffffff', fontsize=10.5, fontweight='bold')
ax.set_xlabel('Ângulo de Flexão do Cotovelo (°)', color='#94a3b8', fontsize=9)
ax.set_ylabel('Torque Exigido no Músculo (N·m)', color='#94a3b8', fontsize=9)
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#255154', alpha=0.5, ls=':')
ax.legend(facecolor='#183739', edgecolor='#255154', labelcolor='#ffffff', fontsize=8)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=200, facecolor=fig.get_facecolor())
plt.close()
```
